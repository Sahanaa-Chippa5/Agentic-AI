## 2. Connect to Ollama
The API key `ollama` is a dummy placeholder for the OpenAI-compatible local API.

````python
from openai import OpenAI
from IPython.display import display, Markdown

ollama = OpenAI(base_url="http://localhost:11434/v1", api_key="ollama", timeout=120.0)
print("Client configured; Ollama must be running locally.")
````

## 3. Set up three LLM personalities

````python
agents = [
    {"name": "Alex", "model": "llama3.2:3b", "personality": "You are optimistic and friendly."},
    {"name": "Blake", "model": "deepseek-r1:1.5b", "personality": "You challenge assumptions respectfully."},
    {"name": "Charlie", "model": "qwen2.5:3b", "personality": "You are logical and neutral."},
]
topic = "Will AI replace software developers?"
conversation = [f"Moderator: Let's discuss: {topic}"]
````

## 4. Call any of the three LLMs
Each model receives its own system prompt plus the full shared conversation as its user prompt.

````python
def call_llm(agent, history):
    system_prompt = (
        f"You are {agent['name']}. {agent['personality']} "
        "You are talking with Alex, Blake, and Charlie. "
        "Reply in 2-3 sentences, stay in character, and do not prefix your name."
    )
    user_prompt = (
        "Conversation so far:\n" + "\n".join(history)
        + f"\n\nNow respond as {agent['name']}. Avoid repeating earlier points."
    )
    response = ollama.chat.completions.create(
        model=agent["model"],
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt},
        ],
    )
    return (response.choices[0].message.content or "").strip()
````

## 5. Run the discussion
Three rounds with three agents = nine local model calls. CPU-only computers may be slow.

````python
for round_number in range(1, 4):
    display(Markdown(f"## Round {round_number}"))
    for agent in agents:
        reply = call_llm(agent, conversation)
        conversation.append(f"{agent['name']}: {reply}")
        display(Markdown(f"**{agent['name']} ({agent['model']}):** {reply}"))
````

## 6. Print the complete conversation

````python
print("\n\n".join(conversation))
````
